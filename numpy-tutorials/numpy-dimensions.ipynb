Working on Dimensions

```python
import numpy as np
```

```python
arr = np.array([100,200,300,400,500,600])
print(arr)
```

```python
np.shape(arr)
```

```python
arr = np.array(
    [
        [100,200,300,400,500,600]
    ]
)
np.shape(arr)
```

```python
arr = np.array(
    [
        [100,200,300],
        [400,500,600]
    ]
)
np.shape(arr)
```

```python
arr = np.array(
    [
        [
            [100,200,300],
            [400,500,600]
        ]
    ]
)
np.shape(arr)
```

```python
arr = np.array(
    [
        [
            [
                [100],[200],[300]
            ],
            [
                [400],[500],[600]
            ]
        ]
    ]
)
np.shape(arr)
```

```python
np.ndim(arr)
```
