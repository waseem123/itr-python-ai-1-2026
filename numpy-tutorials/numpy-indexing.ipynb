Numpy Indexing

```python
import numpy as np
```

```python
arr = np.array(['Java','Python','JavaScript','C++'])
```

```python
print(arr[0])
print(arr[2])
print(arr[1])
print(arr[3])
print(arr[-1])
print(arr[-4])
print(arr[-3])
```

```python
arr = np.array([
            [
                [
                    [['Java','Python','JavaScript']],
                    [['C++','Kotlin','R']],
                    [['C','PHP','Go-lang']]
                ]
            ]
        ])
```

```python
print(arr[0,1,2])
print(arr[0,2,1])
print(arr[0,1,1])
```

```python
print(arr[0,1,0,0])
```

```python
print(arr[0,0,2,0,1])
```

```python
arr = np.array([
            [
                [
                    [['Java','Python','JavaScript']],
                    [['C++','Kotlin','R']],
                    [['C','PHP','Go-lang']]
                ]
            ]
        ])
```

```python
print(arr[-1,-2,-3])
```

```python
print(arr[-1,-1,-3,-3])
```

```python
print(arr[-1,-1,-2,-1,-2])
```

```python
print(arr[0,0,1,0,-2])
```
