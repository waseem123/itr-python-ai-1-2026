```python
import numpy as np
```

```python
arr = np.array(['HTML','CSS','JS','React','Angular','Node.js','jQuery','MongoDB'])
arr
```

```python
print(arr[4])
print(arr[0:])
print(arr[0:5])
print(arr[3:7])
print(arr[:6])
print(arr[2:7:2])
```

```python
arr = np.array(
    [
        ['HTML','CSS','JS','React','Angular','Node.js','jQuery','MongoDB']
    ])
arr
```

```python
print(arr[0,3])
print(arr[0,0:])
print(arr[0,0:5])
print(arr[0,3:])
print(arr[0,1:6])
print(arr[0,0:6])
print(arr[0,0:7:3])
```

```python
arr = np.array(
    [
        [
            ['HTML','CSS','JS','React'],
            ['Angular','Node.js','jQuery','MongoDB']
        ]
     ])
arr
```

```python
print(arr[0,1,2])
print(arr[0,1,2:4])
print(arr[0,1:])
print(arr[0,1,0:3:2])
print(arr[0,0:2,0:2])
```
