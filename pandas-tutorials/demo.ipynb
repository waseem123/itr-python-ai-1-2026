Working with Pandas library

```python
import numpy as np
import pandas as pd
```

Creating Series in Pandas

```python
s1 = np.array([100,200,300,400,500])
s1
```

```python
s = pd.Series(s1)
s
```

```python
s2 = np.array(['C','C++','Python','Java','Javascript','PHP','Kotlin','Dart'])
myseries = pd.Series(s2)
myseries
```

```python
myseries[6]
```

Creating a DataFrame in Pandas

```python
student = {
    'rollno':[101],
    'name':['Roger'],
    'city':['California'],
    'age':[20],
    'marks':[85]
}
```

```python
df = pd.DataFrame(student)
df
```

```python
df['city']
```
