```python
df = pd.read_csv('students.csv')
df
```

```python
df['marks'].fillna(0,inplace=True)
```

```python
df
```

```python
bins = [0, 50, 65, 75, 85, 100]
grades = ['F', 'D', 'C', 'B', 'A']
df['Grade'] = pd.cut(df['marks'], bins=bins, labels=grades,include_lowest=True)
df
```

```python
grades = ['A','B','C','D','F']
plt.pie(df['marks'],labels=df['Grade'],autopct='%.2f%%')
plt.legend()
plt.grid()
plt.show()
```
