```python
df = pd.read_csv('students.csv')
df
```

```python
df['city'].fillna('NA',inplace=True)
df
```

```python
df['marks'].fillna(0,inplace=True)
df
```

```python
avg_age = df['age'].mean()
df['age'].fillna(avg_age,inplace=True)
df
```

```python
df
```

```python
plt.bar(df['name'],df['marks'])
plt.xlabel('Student')
plt.ylabel('Marks')
plt.show()
```

```python
plt.bar(df['rollno'],df['age'],color='purple')
plt.xlabel('Rollno')
plt.ylabel('Age')
plt.grid()
plt.title('Roll Number wise Age of students')
plt.show()
```

```python
plt.barh(df['rollno'],df['age'],color='purple')
plt.xlabel('Age')
plt.ylabel('Rollno')
plt.grid()
plt.title('Roll Number wise Age of students')
plt.show()
```

```python
plt.bar(df['rollno'],df['age'],color='purple',width=0.5)
plt.xlabel('Rollno')
plt.ylabel('Age')
plt.grid()
plt.title('Roll Number wise Age of students')
plt.show()
```

```python
plt.barh(df['rollno'],df['age'],color='purple',height=0.2)
plt.xlabel('Rollno')
plt.ylabel('Age')
plt.grid()
plt.gcf().set_facecolor("#ff00ff39")
plt.gca().set_facecolor("#ff00ff39")
plt.title('Roll Number wise Age of students')
plt.show()
```
