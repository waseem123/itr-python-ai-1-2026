```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([10,20,30,40,50,60])
y = np.array([100,150,250,80,300,190])
```

```python
plt.plot(x,y)
plt.show()
```

```python
plt.plot(x,y,marker='o')
plt.xlabel('X - Axis')
plt.ylabel('Y - Axis')
plt.title('A SAMPLE LINE CHART')
plt.show()
```

A line chart representing rainfall percentage against months

```python
x = np.array(['May','June','July','August','September'])
y = np.array([23,10.5,37.2,5.9,71.8])
```

```python
plt.plot(x,y,marker='*')
plt.xlabel('Months')
plt.ylabel('Rainfall Percentage')
plt.title('Rainfall in Solapur')
plt.grid()
plt.show()
```

```python
plt.plot(x,y,marker='.')
plt.xlabel('Months')
plt.ylabel('Rainfall Percentage')
plt.title('Rainfall in Solapur')
plt.grid()
plt.show()
```

```python
plt.plot(x,y,marker='*',color='#359b21')
plt.xlabel('Months')
plt.ylabel('Rainfall Percentage')
plt.title('Rainfall in Solapur')
plt.grid()
plt.show()
```

```python
plt.plot(x,y,linestyle='--')
plt.xlabel('Months')
plt.ylabel('Rainfall Percentage')
plt.title('Rainfall in Solapur')
plt.grid()
plt.show()
```

```python
plt.plot(x,y,marker='*',color='aqua',linestyle='-.',linewidth='1')
plt.xlabel('Months')
plt.ylabel('Rainfall Percentage')
plt.title('Rainfall in Solapur')
plt.grid()
plt.show()
```

```python
plt.plot(x,y,marker='o',mfc='red',mec='yellow')
plt.xlabel('Months')
plt.ylabel('Rainfall Percentage')
plt.title('Rainfall in Solapur')
plt.grid()
plt.show()
```

```python
plt.plot(x,y,marker='*',markersize='20',mec='black',mfc='yellow')
plt.xlabel('Months')
plt.ylabel('Rainfall Percentage')
plt.title('Rainfall in Solapur')
plt.grid()
plt.show()
```

```python
plt.plot(x,y,marker='*')
plt.xlabel('Months')
plt.ylabel('Rainfall Percentage')
plt.gcf().set_facecolor("#ff00ff39")
plt.gca().set_facecolor("#ff00ff39")
plt.title('Rainfall in Solapur')
plt.grid()
plt.show()
```

```python
plt.plot(x,y,'*-.b',mec='red',mfc='green',markersize=10,linewidth=5)
plt.xlabel('Months')
plt.ylabel('Rainfall Percentage')
plt.gcf().set_facecolor("#ff00ff39")
plt.gca().set_facecolor("#ff00ff39")
plt.title('Rainfall in Solapur')
plt.grid()
plt.show()
```
